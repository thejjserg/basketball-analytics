## 2024-2025 Season Recap

The 2024–25 NBA season represented a major step forward for the Detroit Pistons. After finishing 14–68 (15th in the Eastern Conference) in 2023–24, the team improved to 44–38 (6th in the East) and successfully achieved its primary objective: returning to the Eastern Conference Playoffs.

This leap was driven by the continued emergence of Cade Cunningham as a franchise cornerstone. Cunningham was voted in by Eastern Conference coaches as an All-Star Reserve, earned All-NBA Third Team honors, finished 3rd in the Most Improved Player voting, and placed 7th in MVP voting. His growth was solidified as the team’s offensive engine and late-game decision-maker.

The supporting cast also made meaningful contributions. Malik Beasley finished 2nd for Sixth Man of the Year by providing floor spacing and a scoring punch off-the-bench, while J.B. Bickerstaff was recognized as the Coach of the Year runner-up, reflecting the team’s cultural and structural progress.

From a team identity standpoint, the Pistons established themselves as a physical, defense-first group, finishing with a 113.1 Defensive Rating (11th league-wide). Offensively, the team was closer to league average (115.0 Offensive Rating, 16th), while playing at an above-average tempo (99.8 Pace, 9th). Overall, the Pistons posted a +1.9 Net Rating (13th) — a clear indicator of a competitive playoff-caliber team, but with room to grow.

Detroit finished 4th in the Central Division, behind several strong playoff teams. With significant injuries and player movement affecting division rivals, this upcoming season presents a clear opportunity to capitalize on a favorable competitive window.

The postseason ultimately exposed the team’s relative inexperience. The Pistons were unable to close out the New York Knicks at home, ending our playoff run earlier than we had hoped. While injuries played a role, the series highlighted areas where execution, lineup optimization, and late-game adjustments must improve.

Looking ahead, the roster has undergone meaningful change. The team will move forward without key contributors: Malik Beasley, Tim Hardaway Jr., and Dennis Schröder, but gains stability with the return of Jaden Ivey and Isaiah Stewart from injury. Additionally, the acquisitions of Duncan Robinson and Caris LeVert add shooting versatility and secondary playmaking. Internally, the organization is optimistic about continued development from Jalen Duren, Ausar Thompson, and Ron Holland, each of whom will play a key role in shaping the next phase of the team’s identity.

## Player Profiles

### Cade Cunningham
**Traditional Box Score**
Cade Cunningham finished the 2024–25 season averaging 26.1 PTS, 6.1 TRB, and 9.1 AST per game, while shooting 46.9% from-the-field and 35.6% from three. He functioned as the Pistons’ primary offensive engine and led the team in both scoring and playmaking responsibilities.

**Per 100 Possessions**
Per 100 possessions, Cunningham averaged 35.9 PTS, 8.3 TRB, and 12.5 AST, reinforcing his role as a high-usage initiator rather than a volume scorer operating in isolation. His Offensive Rating (112) and Defensive Rating (113) suggest neutral net impact at the possession level, consistent with a lead guard carrying a heavy offensive burden.

**Advanced Statistics**
Cunningham was the Pistons’ primary shot creator, posting a 33.2% Usage Rate while attempting 23.1 shots per game. His playmaking responsibilities also expanded significantly, as his Assist Percentage jumped to 43%, a career high and a 5.5% increase from the previous season.

As expected for a high-usage guard, Cunningham’s turnover volume was elevated. He averaged 4.4 TOV per game, resulting in a 16% Turnover Rate, the second-highest among NBA players last season. While some of this is an inherent tradeoff of offensive responsibility, reducing live-ball turnovers through improved decision speed and reads against pressure represents a clear area for efficiency gains.

**Adjusted Shooting**
Cunningham produced solid scoring efficiency given his usage, posting a 52.2% eFG% and 56.4% TS%. He was particularly effective at the free-throw line, converting 84.6% of his attempts, and generated consistent pressure on the rim with a 25.4% Free Throw Rate. This combination indicates a scorer who can sustain efficiency even without elite shooting splits, largely due to foul drawing and shot diversity.

**Shooting**
Cunningham’s shot diet (FGA) leaned heavily toward intermediate areas of the floor:
- 16.7% at-the-rim (60.5% FG%)
- 36.2% paint-shots, 3–10 feet (49.7% FG%)
- 12.9% from short, mid-range (10 feet to 16 feet) (47.9% FG%)
- 5.6% from long, mid-range (16 feet to the three-point line) (44.4% FG%)
- 28.7% from three (35.6% 3P%) 

**Play-by-Play**
Cunningham played 84% of his minutes at point guard, clearly functioning as the team’s primary initiator. His On-Court +/- per 100 possessions (+3.4) and On-Off differential (+3.7) indicate that the team performed meaningfully better with him on the floor, reinforcing his value beyond raw box score production.

62.2% of Cunningham’s TOV came from bad passes, highlighting that many mistakes occurred during playmaking rather than ball-handling or scoring attempts. This further supports the idea that streamlining reads and reducing high-risk passes could materially improve offensive efficiency without sacrificing usage.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def plot_shooting_profile_from_excel(
    excel_path: str,
    player_name: str,
    season: str = "2024-25",
    sheet_name: str = "Shooting"
):
    df = pd.read_excel(excel_path, sheet_name=sheet_name)

    required_cols = [
        "Season",
        "%FGA-by-Distance_0-3",
        "%FGA-by-Distance_3-10",
        "%FGA-by-Distance_10-16",
        "%FGA-by-Distance_16-3P",
        "%FGA-by-Distance_3P",
        "FG%-by-Distance_0-3",
        "FG%-by-Distance_3-10",
        "FG%-by-Distance_10-16",
        "FG%-by-Distance_16-3P",
        "FG%-by-Distance_3P",
    ]
    missing = [c for c in required_cols if c not in df.columns]
    if missing:
        raise ValueError(f"Missing expected columns in '{sheet_name}': {missing}")

    df = df.dropna(subset=["Season"]).copy()
    match = df[df["Season"].astype(str) == str(season)]
    if match.empty:
        raise ValueError(
            f"Season '{season}' not found. Seasons available: {sorted(df['Season'].astype(str).unique())}"
        )
    row = match.iloc[0]

    dist_labels = ["0-3", "3-10", "10-16", "16-3P", "3P"]

    # Shot distribution (should sum to ~1.0 already)
    dist_shares = np.array([
        row["%FGA-by-Distance_0-3"],
        row["%FGA-by-Distance_3-10"],
        row["%FGA-by-Distance_10-16"],
        row["%FGA-by-Distance_16-3P"],
        row["%FGA-by-Distance_3P"],
    ], dtype=float)

    # Normalize in case rounding makes sum != 1
    s = np.nansum(dist_shares)
    if s > 0:
        dist_shares = dist_shares / s

    # FG% by distance (decimal form)
    dist_fg = np.array([
        row["FG%-by-Distance_0-3"],
        row["FG%-by-Distance_3-10"],
        row["FG%-by-Distance_10-16"],
        row["FG%-by-Distance_16-3P"],
        row["FG%-by-Distance_3P"],
    ], dtype=float)

    # ----------------------------
    # Chart 1: PIE (FGA split)
    # ----------------------------
    plt.figure(figsize=(8, 6))
    plt.pie(
        dist_shares,
        labels=dist_labels,
        autopct=lambda p: f"{p:.1f}%"
    )
    plt.title(f"{player_name} FGA")
    plt.show()

    # ----------------------------
    # Chart 2: BAR (FG% by distance)
    # ----------------------------
    plt.figure(figsize=(8, 5))
    x = np.arange(len(dist_labels))
    plt.bar(x, dist_fg)
    plt.xticks(x, dist_labels)
    plt.ylim(0, 1)
    plt.ylabel("FG% (decimal)")
    plt.title(f"{player_name} FG%")
    plt.show()

    # Optional: quick printed summary row for debugging/QA
    summary = pd.DataFrame({
        "Distance": dist_labels,
        "%FGA": dist_shares,
        "FG%": dist_fg
    })
    print(f"Summary used for charts ({player_name}, {season}):")
    print(summary.to_string(index=False))
```

```python
def plot_player_trend(
    excel_file,
    stat_column,
    y_label,
    title,
    x_column="Season",
    first_season=None,
    last_season=None,
    y_min=0,
    y_max=None,
    filename=None
):
    import os
    import pandas as pd
    import matplotlib.pyplot as plt

    # Read data
    df = pd.read_excel(excel_file, sheet_name="Shooting")

    # Create numeric season key
    df["Season_Start"] = df["Season"].str[:4].astype(int)

    # Apply season filters
    if first_season is not None:
        df = df[df["Season_Start"] >= first_season]

    if last_season is not None:
        df = df[df["Season_Start"] <= last_season]

    # Sort correctly
    df = df.sort_values("Season_Start")

    # Handle percentages
    y_values = df[stat_column]
    if y_values.max() <= 1:
        y_values = y_values * 100

    # Plot
    plt.figure(figsize=(10, 6))
    plt.plot(df[x_column], y_values, marker="o", linewidth=2)

    plt.xlabel(x_column)
    plt.ylabel(y_label)
    plt.title(title)
    plt.ylim(bottom=y_min)

    if y_max is not None:
        plt.ylim(top=y_max)

    plt.grid(True, alpha=0.3)

    # Save
    if filename:
        os.makedirs("Charts", exist_ok=True)
        plt.savefig(f"Charts/{filename}", dpi=300, bbox_inches="tight")

    plt.show()
```

```python
players = ["Cade Cunningham", "Duncan Robinson", "Ausar Thompson", "Tobias Harris", "Jalen Duren", "Isaiah Stewart", "Jaden Ivey", "Ron Holland"]
for player in players:
    plot_player_trend(
    excel_file=(player + ".xlsx"),
    stat_column="FG%-by-Distance_3P",
    y_label="3P%",
    title= player + " – 3PT Shooting Trend",
    first_season=2020,
    last_season=2024,
    y_min=0,
    y_max=45,
    filename=player + "_3P_Trend_2021_2024.png"
)
```
