## Dashboards & Visual Tools
### What a Dashboard Is and Isn't

- **Dashboard** = Monitor + Diagnose + Act - Not a KPI Graveyard
- Built for a specific user + decision cadence (daily, game-to-game, scout cycle, etc.)
- Answers a primary question per view
- Pair KPIs with drivers and definitions so people trust what they see

### Dashboard Anatomy - 3 Layers
- Layer 1 - KPIs
  - KPIs for aligned goals (OER / DER, Four Factors, Shot Diet)
- Layer 2 - Drivers
  - Shot Quality vs Shot Making
  - Turnover Types
  - Foul Profile
- Layer 3 - Actions
  - Matchup Keys
  - Play Calls
  - Individual Players

### Designing for the User
- **Coach View**: Fewer metrics, faster scanning, clear action items
- **Front Office View**: More context, variance, segments, and long-term trends
- Minimize the scroll
- Tabs vs Too Much Data on One-Page

### Filters & Drill Paths
- Core Filters:
  - Date Range
  - Game Range
  - Player Played vs Player Didn't Play
  - Player Start vs Player Bench
  - Game State: Time and Score
- Drill Path:
  1. Team
  2. Lineup
  3. Player
- Always Display Active Filters

### Why Dashboards Fail
- Too many metrics (no hierarchy)
- Not tied to decisions
- Updates too slowly
- Definitions change or aren't visible
- Hard to scan or inconsistent formatting

## Visualizations for Player, Team, and League Insights
### Player Visuals
- Role context + volume + efficiency + baseline / percentile
- Always show sample size (minutes, attempts, possessions)
- Separate shot quality from shot making when possible (expected vs actual)
- Relevant metrics for position / role


### Player Chart Patterns that Work
- Shot profile: zone share + efficiency + expected (as separate layers)
- xPPS vs PPS scatter
- Strength / Weakness Table
- Small multiples for before-and-after or role-based splits


### Team Visuals: Diagnosing Offense / Defense
- Start from posessions & outcomes
- Efficiency / Four Factors relative to the league
- Shot profile & quality presented separately (avoid *one mega chart*)
- Turnovers by Type / Shot Clock
- Rebounding: Chances vs Conversions
- Exmaple: [Team Efficiency Visualization](https://x.com/kirkgoldsberry/status/1905609850224026054)

### Lineups & Rotations
- Rotation Timeline View: Who plays with whom and when?
- Lineup Summaries: Identity (Efficiency / Four Factors / Shot Diet)
- Use Minimum-Minute rules and label them on the slide

### League Context and Comparisons
- Percentiles by position or role (avoid *all-player* comparisons)
- Similarity comps with guardrails (minutes, role, usage)
- Show Both: where the player ranks and what the player actually does

## Data Visualization Best Practices
### Fonts and Labeling
- Create a clear hierarchy
- Limit font families and keep styles consistent
- Prefer direct labels over legends when possible
- Avoid hard to read text rotations (ex: upside down)
- Don't go too small - screen vs print sizing matters

### Color Choices
- Categorical: Distinct Hues for Discrete Groups
- Sequential:
  - Light to Dark for Magnitude
  - One Color Scale
- Diverging: Two color scale around a meaningful center (0, League Average)
- Use one accent color for emphasis; mute everything else
- Check color-vision accessibility and contrast on projector screens or TVs

### Number Alignment and Formatting
- Right-align numbers, center headers
- Bottom-aligned in cell
- Standardize decimals, units, and rounding rules
- Use sign conventions consistently
- Show denominator context (attempts, possessions, minutes)
- Keep columns consistent sizes
- Clear purpose to order (time-series, descending / ascending)

### Building Consistent Charts
- Use a template
- Same metric should look the same everywhere
- Standard titles
- Create *tokens*: colors, label styles

### Details that Signal Credibility
- Margins + Grid: Consistent whitespace improves scanning
- Footer: Date range, opponent / season context, page number date produced
- Source / Definitions
- Confidentiality disclaimer when appropriate

## Storytelling: Charts, Graphs, & Data Narratives
### A Simple Data Story Framework
- Context: What *normal* looks like - a baseline + a timeframe
- Tension: What changed? What's different from normal?
- Mechanism: The driver supported by 1-2 visuals
- Decision: What action follows; how do we measure success?

### Titles & Annotations that Drive Action
- Title: The Takeaway
- Subtitle: Scope
- Annotate only where it changes interpretation
- Put definitions / notes in the footer

### One Slide Story Template + Close
- One Side: Visual evidence (one main chart)
- Other side: 2-3 action bullets
- Footer: Sample, date range, source, definitions

## Final Project: Analytics-Driven Coaching or Front Office Strategy
### **Analytics-Driven Coaching**: 
Develop a data-driven strategy for how to best utilize an NBA roster for the following season.

**GOAL**: Figure out what to prioritize and where to prioritize strategic initiatives for an NBA team's roster to provide to a Head Coach. Focus on factors like *optimizing shot selection*, *shooting efficiency*, *ball control*, *rebounding*, and *free throw performance* to maximize wins to fit a current NBA team.

Make sure to include the following:
1. **Data Collection** from either or both [NBA Stats]() or [Basketball Reference]()
2. **Player Performance**: Evaluate individual player metrics to identify individual tendencies in the context of the team. Focus on creating reports on the top 8 players you would recommend playing in this team's rotation.
3. **Team General Strategy**: Note tactical and strategic adjustments based on optimizing your roster, whether those are consistent game-to-game or dynamic
4. **Lineup Analysis**: Assess combinations of players that have played together previously, as well as those that have not based on their demonstrated skills and metrics. Provide recommendations of 5-man lineups that would project to perform well as well as other combinations of players that you would recommend should or should not share the court.
5. **Game Specific Adjustments**: Where will your team struggle? How will you have to adjust to certain styles or opponent rosters?
6. **Visualizations**: Summarize important areas of data for this team in the form of static or dynamic visualizations.


### **Front Office Strategy**:
Develop a data-driven roster strategy, incorporating evaluating player performance, draft prospects, free agents, and optimizing salary cap management for short- and / or long-term success.

**GOAL**: Design a comprehensive roster plan that balances current performance, potential development, and salary cap efficiency to provide to a GM. Focus on factors like *maximizing player value relative to the market*, *prospect scouting*, *roster construction*, and *strategic planning*. 

Make sure to include the following:
1. **Data Collection** from either or both [NBA Stats]() or [Basketball Reference]()
2. **Player Valuation**: Assess the current roster utilizing basic and advanced metrics and provide recommendations for 3 Free Agents the team should look to sign
3. **Prospect Scouting**: Based on this team's positioning in the draft, analyze a draft prospect's fit with the team's needs and potential impact
4. **Roster Construction**: Evaluate two realistic trade opportunities and the projected impact on team performance and salary cap implications. Include reasoning of why this would benefit the trade partner to help both teams accomplish their goals
5. **Strategic Planning**: Illustrate a 3-Year Roadmap based on the status of the roster after following your recommendations
6. **Visualizations**: Present key insights in dashboards, charts, and graphs to simplify data
