## Translating Complex Statistical Concepts
### Analysis into Action
1. Start with your claim
2. Back it up with evidence
3. Provide supporting visuals

#### From Stats to Stakes
- Lead with the problem in basketball language, then support with numbers
- Convert...
  - Shot Quality, **Good Looks** vs **Bad Shots**
  - Lineup Splits, **This group wins because...**
  - Model Output, **Here is what we can control...**
- ... Into Stakes & Behavior
  - For example, *You're creating value getting to the rim. We want more of that.*
  - For example, *Survives with rebounding & not turning it over, struggles vs switching.*

#### Uncertainty Without Sounding Unsure
- Frame signal vs noise: stable trends vs noisy outcomes
- Use ranges or tiers instead of false precision
- Explicit with context: Sample, opponent lineups, role changes
- *If not, then what?*

## Tailoring Communication
### Players: Impact & Development
- Lead with impact - How do you help the team?
- Connect to personal motivations
  - How does this get the player more minutes?
  - More points?
  - More money?
  - More legacy impact?
- Start with behavior, not a metric
- One number to back it up
- Make it executable - *When you see X, then do Y*
  - Define Success
 
### Communication: 2-2-2 Method
- **Explain**
  - Big picture, what can they control
  - Explain the basics of the concept
  - Tie to impact & development
  - Keep it in accessible language0
- **Show**
  - 1 visual and / or a few clips
  - Before-and-After comparison
  - Incorrect vs Correct
  - Keep the messaging simple and consistent
- **Clarify**
  - Reiterate what success is
  - Open floor for questions and clarifications
  - End with clear takeaways

### Coaches: Direct + Simple
- Lead with the recommendation, not the setup
- Use intuitive language whenever possible; numbers to support the point
- Support with 2-3 bullets:
  - Why it works?
  - When it fails, then what are the implementation details?
- Add the counter example: In the scenario the recommendation fails, then what's Plan B?
- Make it **operational**, How does this impact a decision?


### Front Office: Decision-Ready EV + Risk
- Quantify expected impact as a range - wins, efficiency, availability, role value, etc.
- State costs & constraints
  - Relative to the cap
  - Roster fit
  - Minutes
- Surface risk & assumptions
- Give a timeline & next steps

### Consistent Templates
- Your reports should feel consistent to the receiver
- Information displayed / communicated consistently = higher likelihood of adoption
  - Less time finding the information
  - More comfort when looking at the report
  - Easier to understand the messaging

## Building Trust & Buy-In
### Trust Stack: What Earns Belief
- Shared definitions
  - Best metrics to use
  - Understanding of metrics used. For example, what does TS% actually measure? What is a good or bad number?
- Transparency
- Mechanism to Direct Impact
- Consistency
- Respect

### Handling Pushback
- Scripts you can use
  - *That is fair - here's what would change my recommendation...*
  - *Two assumptions: **A** & **B**. If either are wrong, then we can pivot to **C**.*
  - *We have a limited sample. We'll continue to monitor and validate the next few games.*
  - *If they counter with **X**, then our answer is **Y**. Here is the trigger...*

## Workshop: Role-Play a Presentation to a Coach or General Manager
1. Choose an audience: **Coach** or **GM**
2. 2:30 minutes **MAX** presentation, then a follow-up question
3. Title, a recommendation (not the topic)
4. One visual (clean labels; Class 12 Standards)
5. 3 Bullets Max
6. One *Now What* action line
7. Guardrail Line - sample, context, or uncertainty point

- Pick one scenario to role-play
  1. Rotation or Lineup Tweak
  2. Shot Profile Emphasis
  3. Player Development Point of Emphasis for One Player

## The Point of Communication
- Dashboards are for **explorations**. Meetings are for **decisions**.
- Start with the claim
- Use the least amount of evidence needed to make your point, but don't fall short of proving your statement.
- Tailor to the decision-maker
- Earn buy-in with transparency and guardrails
